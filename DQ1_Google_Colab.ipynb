## Part A

````python
# ERROR 1 (ModuleNotFoundError): Incorrect library name 'panda'. FIX: Changed 'panda' to 'pandas'.
import pandas as pd
````

## Part B

````python
# ERROR 2 (ValueError): Worksheet named 'Order' not found. FIX: Omitted the sheet_name parameter to load the first sheet automatically.
df = pd.read_excel("DQ1_Datasheet.xlsx")
````

## Part C

````python
print(df.head())
# ERROR 3 (TypeError): 'tuple' object is not callable when using df.shape(). FIX: Removed the parentheses because shape is an attribute, not a method.
print("Rows and columns:", df.shape)
# ERROR 4 (AttributeError): 'DataFrame' object has no attribute 'column'. FIX: Changed 'column' to 'columns'.
print("Columns:", list(df.columns))
````

## Part D

````python
# ERROR 5 (KeyError): 'price' column name is case-sensitive. FIX: Changed 'price' to 'Price'.
df["Amount"] = df["Price"] + df["Quantity"]
df["Total"]  = df["Amount"] + df["Amount"] * 18 / 100
df["Boxes"]  = df["Quantity"] // 3
df["Left"]   = df["Quantity"] % 3
print(df[["Order_ID", "Amount", "Total", "Boxes", "Left"]])
````

## Part E

````python
slow_orders  = (df["Delivery_Days"] > 5).sum()
# ERROR 6 (ValueError/LogicalError): Filter condition for 'rating 4 or more' used strictly greater than (>) instead of greater than or equal to (>=). FIX: Changed '>' to '>='.
good_ratings = (df["Rating"] >= 4).sum()
print("Slow deliveries (more than 5 days):", slow_orders)
print("Orders with rating 4 or more     :", good_ratings)
````

## Part F

````python
for i in range(len(df)):
    order_id = df["Order_ID"][i]
    days     = df["Delivery_Days"][i]
    rating   = df["Rating"][i]

    if days <= 2:
        status = "Fast"
    # ERROR 7 (SyntaxError): Missing colon ':' at the end of the elif condition. FIX: Added ':' to complete the syntax.
    elif days <= 5:
        status = "Normal"
    else:
        status = "Slow"

    if days > 5:
        if rating >= 2:
            action = "Call customer"
        else:
            action = "Send apology SMS"
    else:
        action = "No action"

    print(order_id, "|", status, "| Rating", rating, "|", action)
````
